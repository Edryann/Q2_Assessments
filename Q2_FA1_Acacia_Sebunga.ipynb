```python


num1 = float(input("Enter First Number: "))
num2 = float(input("Enter Second Number: "))

print("Select an operation:")
print("1 Addition")
print("2 Subtraction")
print("3 Multiplication")
print("4 Division")

operation = input("Enter Choice (1, 2, 3, or 4): ")

if operation == "1":
    result = num1 + num2
    print("The answer is:", result)

elif operation == "2":
    result = num1 - num2
    print("The answer is:", result)

elif operation == "3":
    result = num1 * num2
    print("The answer is:", result)

elif operation == "4":
    result = num1 / num2
    print("The answer is:", result)

else:
    print("Invalid choice! Please run the program again and pick 1, 2, 3, or 4.")
```

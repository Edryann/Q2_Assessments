```python


def add_numbers(num1, num2):
    return num1 + num2

def subtract_numbers(num1, num2):
    return num1 - num2

def multiply_numbers(num1, num2):
    return num1 * num2

def divide_numbers(num1, num2):
    if num2 == 0:
        return "Error! Division by zero."
    return num1 / num2

num1 = float(input("Enter First Number: "))
num2 = float(input("Enter Second Number: "))

print("\nSelect an operation:")
print("1 Addition")
print("2 Subtraction")
print("3 Multiplication")
print("4 Division")

operation = input("Enter Choice (1, 2, 3, or 4): ")

if operation == "1":
    result = add_numbers(num1, num2)
    print("The answer is:", result)
elif operation == "2":
    result = subtract_numbers(num1, num2)
    print("The answer is:", result)
elif operation == "3":
    result = multiply_numbers(num1, num2)
    print("The answer is:", result)
elif operation == "4":
    result = divide_numbers(num1, num2)
    print("The answer is:", result)
else:
    print("Invalid choice! Please run the program again and pick 1, 2, 3, or 4.")


#REFLECTION
#1, What functions did you create in your calculator program?
#I created functions such ass addition, subtraction, multiplication and division.

#2, What parameters did your functions use?
# my parameters were num1 and num2.

#3, What arguments were passed when the functions were called?
# The arguments passed into the functions were the variables num1 and num2

#4, How did your program use the returned value?
# The program captured the returned value by assigning it to a variable named result.

#5, Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?
# Using functions makes the code reusable and much cleaner to read than a single long block of code.
```
